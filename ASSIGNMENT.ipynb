```python
#row wise and column wise sum
import pandas as pd
import numpy as np

data = np.random.randint(1, 20, (4, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("Row-wise Sum:")
print(df.sum(axis=1))

print("Column-wise Sum:")
print(df.sum(axis=0))










```

```python
#Fill NaN Values with Column Mean
import pandas as pd
import numpy as np

data = np.random.randint(1, 20, (5, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[4, 0] = np.nan

print("Before:")
print(df)

df = df.fillna(df.mean())

print("After:")
print(df)
```

```python


#Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

```python
#Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Bottle'],
    'Category': np.random.choice(['A', 'B', 'C'], 6),
    'Sales': np.random.randint(100, 1000, 6)
 })

print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total Sales:")
print(result)
```

```python
#Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ayaan', 'Rahul', 'Ali']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 80]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

```python
#Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('A', 1),
    ('A', 2),
    ('B', 1),
    ('B', 2)
], names=['Category', 'Number'])

df= pd.DataFrame({

   'Value': [10, 20, 30, 40]
}, index=index)

print("DataFrame:")
print(df)

print("Access A:")
print(df.loc['A'])

print("Access A, 1:")
print(df.loc[('A', 1)])

print("Slicing:")
print(df.loc['A':'B'])
```

```python
#. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B', 'C', 'C'],
    'SubCategory': ['X', 'Y', 'X', 'Y', 'X', 'Y'],
    'Value': np.random.randint(10, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])

print(df)

print("Sum of Values:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
#Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame
import pandas as pd
import numpy as np

data = np.random.randint(1, 20, (5, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.map(lambda x: x * 2)

print("Doubled DataFrame:")
print(result)
```

```python
# Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.
import pandas as pd
import numpy as np

data = np.random.randint(1, 20, (6, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print(df)
```
